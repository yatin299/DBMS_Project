# WS3: Telco Customer Churn - Data Preprocessing & Visualization

**Task 1 - Load the Dataset**

```python
import pandas as pd

df = pd.read_csv("/content/WA_Fn-UseC_-Telco-Customer-Churn.csv")
df.head()
```

**Task 2 - Understand the Structure**

```python
print(df.shape)
print(df.info())
print(df.describe())
```

**Task 3 - Handle Missing and Incorrect Data**

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
df.dropna(inplace=True)
df.isnull().sum()
```

**Task 4 - Drop Unnecessary Columns**

```python
if 'customerID' in df.columns:
    df.drop(['customerID'], axis=1, inplace=True)
df.head()
```

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
df.dropna(inplace=True)
```

```python
 if 'customerID' in df.columns:
    df.drop(['customerID'], axis=1, inplace=True)
 df
```

**Task 5 - Standardize Categorical Variables**

```python
df.replace({'No internet service': 'No', 'No phone service': 'No'}, inplace=True)
df
```

```python
 df.columns
```

**Task 6 - Encode Categorical Variables**

```python
from sklearn.preprocessing import LabelEncoder

binary_cols = ['gender', 'Partner', 'Dependents', 'PhoneService', 'PaperlessBilling', 'Churn']
le = LabelEncoder()
for col in binary_cols:
    df[col] = le.fit_transform(df[col])

multi_cols = ['MultipleLines', 'InternetService', 'OnlineSecurity', 'OnlineBackup',
              'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies',
              'Contract', 'PaymentMethod']
df = pd.get_dummies(df, columns=multi_cols)
df
```

**Task 7 - Feature Scaling**

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df[['tenure', 'MonthlyCharges', 'TotalCharges']] = scaler.fit_transform(df[['tenure', 'MonthlyCharges', 'TotalCharges']])
df.head()
```

**Task 8 - Univariate Visualizations**

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set(style="whitegrid")

# Histograms with customized colors
for col, color in zip(['tenure', 'MonthlyCharges', 'TotalCharges'], ['skyblue', 'salmon', 'green']):
    plt.figure(figsize=(6, 4))
    sns.histplot(df[col], kde=True, bins=30, color=color)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()

# Countplots
for col in ['gender', 'Churn']:
    plt.figure(figsize=(6, 4))
    sns.countplot(data=df, x=col)
    plt.title(f'Countplot of {col}')
    plt.xticks(rotation=0)
    plt.show()

for col in ['InternetService_DSL', 'InternetService_Fiber optic', 'InternetService_No']:
    plt.figure(figsize=(6, 4))
    sns.countplot(data=df, x=col, hue=col, legend=False)
    plt.title(f'Countplot of {col}')
    plt.xticks(rotation=0)
    plt.show()

for col in ['Contract_Month-to-month', 'Contract_One year', 'Contract_Two year']:
    plt.figure(figsize=(6, 4))
    sns.countplot(data=df, x=col, hue=col, legend=False)
    plt.title(f'Countplot of {col}')
    plt.xticks(rotation=0)
    plt.show()
```

**Task 9 - Bivariate Analysis**

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set(style="whitegrid")

# Boxplot
plt.figure(figsize=(6, 4))
sns.boxplot(data=df, x='Churn', y='MonthlyCharges', palette='coolwarm')
plt.title('Monthly Charges by Churn')
plt.show()

# Countplot
plt.figure(figsize=(8, 6))
sns.countplot(data=df, x='Contract_Month-to-month', hue='Churn', palette='husl')
plt.title('Churn by Month-to-month Contract')
plt.xticks(rotation=0)
plt.show()

plt.figure(figsize=(8, 6))
sns.countplot(data=df, x='Contract_One year', hue='Churn', palette='husl')
plt.title('Churn by One Year Contract')
plt.xticks(rotation=0)
plt.show()

plt.figure(figsize=(8, 6))
sns.countplot(data=df, x='Contract_Two year', hue='Churn', palette='husl')
plt.title('Churn by Two Year Contract')
plt.xticks(rotation=0)
plt.show()

# Correlation heatmap
corr = df.select_dtypes(include='number').corr()
plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Correlation Heatmap')
plt.show()
```

**Task 10 - Multivariate Visualizations**

```python
# Pairplot
sns.pairplot(df[['tenure', 'MonthlyCharges', 'TotalCharges', 'Churn']],
             hue='Churn', palette='husl')
plt.suptitle('Pairplot of Selected Features by Churn', y=1.02)
plt.show()

# Barplot
plt.figure(figsize=(6, 4))
sns.barplot(data=df, x='InternetService_Fiber optic', y='TotalCharges', hue='Churn')
plt.title('Avg TotalCharges by InternetService (Fiber Optic) and Churn')
plt.xticks(rotation=0)
plt.show()
```

**Task 11 - Time-based Visualization**

```python
df['tenure_group'] = pd.cut(df['tenure'],
                            bins=[0, 12, 24, 36, 48, 60, 72],
                            labels=['0-12','12-24','24-36','36-48','48-60','60-72'])

plt.figure(figsize=(7, 4))
sns.countplot(data=df, x='tenure_group', hue='Churn', palette='Set1')
plt.title('Churn Distribution Across Tenure Groups')
plt.show()
```

Task 12 - Feature Engineering

```python
# Service columns after encoding
service_cols_encoded = ['MultipleLines_Yes', 'PhoneService', 'InternetService_DSL',
                        'InternetService_Fiber optic', 'OnlineSecurity_Yes',
                        'OnlineBackup_Yes', 'DeviceProtection_Yes', 'TechSupport_Yes',
                        'StreamingTV_Yes', 'StreamingMovies_Yes']

# Calculate total services by summing the encoded columns
df['TotalServices'] = df[service_cols_encoded].sum(axis=1)

# Assuming 'SeniorCitizen' is already encoded as 0/1
# Assuming 'Partner' is already encoded as 0/1
df['IsSeniorOrPartner'] = df.apply(
    lambda row: 1 if row['SeniorCitizen'] == 1 or row['Partner'] == 1 else 0, axis=1
)

df[['TotalServices', 'IsSeniorOrPartner']]
```
